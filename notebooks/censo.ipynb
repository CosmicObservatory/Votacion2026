# Descarga de Datos del Censo

```python
from vothmo.config import CENSO_DATA, censo_url, FOLDER_RAW
import os
import requests as r

files = ["2019/09/DatosAbiertos_DECEyEC_ConteosCensales_PrisionPreventiva2024-1.zip",
"2019/09/DatosAbiertos_DECEyEC_ConteosCensales_VotoAnticipado2024.zip",
"2019/09/DatosAbiertos_DECEyEC_ConteosCensales2024.zip",
"2022/04/DatosAbiertos_DECEyEC_ConteosCensales2021.zip",
"2022/04/DatosAbiertos_DECEyEC_ConteosCensales2018.zip",
"2022/04/DatosAbiertos_DECEyEC_ConteosCensales2015.zip",
"2022/04/DatosAbiertos_DECEyEC_ConteosCensales2012.zip",
"2022/04/DatosAbiertos_DECEyEC_ConteosCensales2009.zip"
]

for f in files:
    filename = f.split("/")[-1]
    link = censo_url / f
    if not os.path.exists(FOLDER_RAW / filename):
        response = r.get(link)
        response.raise_for_status()
        with open(FOLDER_RAW / filename, "wb") as file:
            file.write(response.content)
```

# Extracción del Censo

```python
from zipfile import ZipFile

for f in os.listdir(FOLDER_RAW):
    name = f.split(".")[0]
    path = FOLDER_RAW / f
    if not os.path.exists(CENSO_DATA/name):
        with ZipFile(path, "r") as zp:
            zp.extractall(path=CENSO_DATA/name)
```
