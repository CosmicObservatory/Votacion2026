# Analisis de Datos del Censo Estatal.

Una vez los datos estan descargados. Procedemos con un analisis de los datos.


```python
from vothmo.config import CENSO_DATA
import os

data_folders = os.listdir(CENSO_DATA)

files = {}

for folder in data_folders:
    fdr = CENSO_DATA / folder
    lfdr = os.listdir(fdr)
    folder_pth = fdr / lfdr[0]
    for file in os.listdir(folder_pth):
        files[file] = folder_pth / file

files
```

Vemoos una mezcla de archivos .txt y .csv. Entonces, vamos a estructurarlos para un mejor analisis.

```python
txt_files = [f for f in files.keys() if f.endswith(".txt")]
txt_files
```

## Diccionario de datos.

Notemos que los archivos .txt corresponden a los diccionarios de datos. Entonces debemos extraer la información relevante.

```python
d0 = files[txt_files[-1]]

with open(d0, "r") as dic:
    text = dic.read()

codebook = text.split("\n")
vart = {}
adinfo = {}
k = 0
for c in codebook:
    line = c.split("\t")
    if len(line) == 2:
        if line[0] != "":
            vart[line[0]] = line[1]
        else:
            adinfo[k] = line[1]
    k+=1
# Extraemos las variables relevantes
vart
```

Notemos que adinfo extrajo los motivos del valor NS. Sin emabargo, esto no es relevante para nuestro analisis. Entonces extraeremos los codigos relevantes.

```python
states= {}
codes = {}
for k, v in vart.items():
    try:
        int(k)
        states[k] = v
    except:
        codes[k] = v

states
```

### Conglomerado

Finalmente haremos un conglomerado del diccionario de datos. Extrayendo la información de los estados y las variables.

```python
import pandas as pd

datadict = pd.DataFrame(index = codes.keys(),
                        data= codes.values(),
                        columns=['Descripción'])

mexico = pd.DataFrame(index= states.keys(),
                        data=states.values(),
                        columns=["Descripción"])
display(datadict)
mexico
```

Finalmente, haremos el conjunto de datos por archivo

```python

def codebook(file):
    with open(file, "r") as dic:
        text = dic.read()

    codebook = text.split("\n")
    vart = {}
    adinfo = {}
    k = 0
    for c in codebook:
        line = c.split("\t")
        if len(line) == 2:
            if line[0] != "":
                vart[line[0]] = line[1]
            else:
                adinfo[k] = line[1]
        k+=1
    
    states= {}
    codes = {}
    for k, v in vart.items():
        try:
            int(k)
            states[k] = v
        except:
            codes[k] = v

    datadict = pd.DataFrame(index = codes.keys(),
                        data= codes.values(),
                        columns=['Descripción'])

    mexico = pd.DataFrame(index= states.keys(),
                            data=states.values(),
                            columns=["Descripción"])
    return datadict, mexico, adinfo

import json

for f in txt_files:
    codes, maps, adinfo = codebook(files[f])
    name = f.split("_")[-1][:-4]
    codes.to_csv(CENSO_DATA / f"code_{name}.csv")
    maps.to_csv(CENSO_DATA / f"states_{name}.csv")
    if len(adinfo) > 0:
        print(adinfo)
        with open(CENSO_DATA / f"info_{name}.json", "w") as info:
           json.dump(adinfo, info)
```
