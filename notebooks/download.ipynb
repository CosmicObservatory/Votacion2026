Comenzaremos con la descarga de los resultados electorales por distrito 


## Descarga de los resultados de las elecciones.

```python
from vothmo.config import DATA_DIR
from yarl import URL
import requests as r
from bs4 import BeautifulSoup

# Acceso a la pagina.
iee_url = URL("https://ieesonora.org.mx/resultados/computo/diputaciones")

response = r.get(iee_url)
response.raise_for_status()

# Filenames
soup = BeautifulSoup(response.content, "lxml")

file_ref = soup.find("select")
files = file_ref.find_all("option")

filename = [f"{f.text}.xlsx" for f in files if f.text != ""]

# Normalize Filenames
norm_names = []
for name in filename:
    try:
        name.index("Ñ")
        name = name.replace("Ñ","N")
        norm_names.append(name)
    except:
        try:
            name.index("Ó")
            name= name.replace("Ó","O")
            norm_names.append(name)
        except:
            norm_names.append(name)

norm_names
```

## Por Distrito

### Elecciones 2018,2021

```python
year = [2018, 2021]

folder = DATA_DIR / "raw" / "Distrito"
for y in year:
    for file in filename:
        download_url = URL(f"https://ieesonora.org.mx/_elecciones/procesos/{y}/resultados/basedatos/")
        name = file.replace(" ", "_") if y==2018 else file
        down_req = download_url / name
        savepath = folder / f"{y}"
        response = r.get(down_req)
        response.raise_for_status()
        with open(savepath / name, "wb") as distrito:
            distrito.write(response.content)
```

### Elecciones 2024

```python
download_url= URL("https://www.ieesonora.org.mx/elecciones2024/computo/CDE/")
savepath = folder  / "2024"
for name in filename:
    down_req = download_url / name
    response = r.get(down_req)
    response.raise_for_status()
    with open(savepath / name, "wb") as distrito:
        distrito.write(response.content)
```

### Información Poblacional.

```python
map_dist = URL("https://www.ieesonora.org.mx/documentos/estadistica_cartografia/distrito/")
folder = DATA_DIR / "raw" / "Distrito"
for i in range(1, 22):
    filename = f"distrito{i}.xls"
    down_req = map_dist / filename
    savepath = folder / "Mapas"
    response = r.get(down_req)
    response.raise_for_status()
    with open(savepath / filename, "wb") as distrito:
        distrito.write(response.content)
```
