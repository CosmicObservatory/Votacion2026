```python
import geopandas
from datetime import date
from pathlib import Path
import requests as r
from bs4 import BeautifulSoup
import os
import yarl
from vothmo.config import DATA_DIR


INE_mapas = "https://pautas.ine.mx/transparencia/mapas"
maps = yarl.URL(INE_mapas)
year_limit = date.today().year
year_start = 2020
years = list(range(year_start, year_limit + 1))
pages = [f"index_{y}.html" if y != year_limit else "index.html" for y in years]
download_pages = []
for p in pages:
    link = maps / p
    response = r.get(link)
    response.raise_for_status()

    soup = BeautifulSoup(response.content, "lxml")

    links = soup.find_all("a")
    useful_links = []
    for l in links:
        if l.text=="Descargar archivo":
            useful_links.append(l['href'])
    print(useful_links)
    download_pages += useful_links
download = set(download_pages)
```

```python
download_req = [d for d in download if not d.split("/")[1].startswith("Area")]
download_req
```

```python
folder = DATA_DIR / "raw" / "Mapas"

for fd in download_req:
    year = fd.split("/")[0].split("_")[-1]
    filename = fd.split("/")[1]
    if not os.path.exists(folder / f"{year}"):
        os.mkdir(folder / f"{year}")
    dl = maps / fd
    response = r.get(dl)
    response.raise_for_status()
    with open(folder / f"{year}" / filename, "wb") as zpf:
        zpf.write(response.content)
```
