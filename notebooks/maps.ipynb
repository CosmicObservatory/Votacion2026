```python
import requests

INE_mapas = "https://pautas.ine.mx/transparencia/mapas"
maps = yarl.URL(INE_mapas)
year = [2020, 2021, 2022, 2023, 2024, 2025, 2026]
sectors = ["ENTIDAD", "MUNICIPIO",
            "DISTRITO_FEDERAL", "DISTRITO_LOCAL",
            "SECCION", "LOCALIDAD", "MANZANAS",
            "LOCALIDADES_URBANAS", "LOCALIDADES_RURALES"]

def get_download_link(year, name, url=maps):
    url_year = f"cob_{year-1}"
    filename = f"{name}.zip"
    return url / url_year / filename

headers = {
    "User-Agent": "Mozilla/5.0 (X11; Linux x86_64; rv:154.0) Gecko/20100101 Firefox/154.0"
}

for y in year:
    for s in sectors:
        link = get_download_link(y, s)
        r = requests.get(link, headers=headers)
        if r.status_code != 200:
            print(link)
            print("Fallo ", (y, s))
        else:
            print("EXITO", (y, s))
```
